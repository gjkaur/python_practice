## 1. Core Built-in Types (PCEP 1.3)

| Type | Description | Examples |
|------|--------------|----------|
| **bool** | Boolean logic | `True`, `False` |
| **int** | Arbitrary-precision integer | `0`, `42`, `-10` |
| **float** | Double-precision floating-point | `0.0`, `3.14`, `1.5e3` |
| **str** | Text (sequence of Unicode code points) | `"hello"`, `'world'` |

### How to inspect type

- **`type(x)`** – returns the type object of `x`.
- **`isinstance(x, T)`** – returns `True` if `x` is an instance of type `T` (or a subclass).

```python
is_active = True
user_count = 120
discount_rate = 0.15
welcome_message = "Welcome"

print(type(is_active), type(user_count), type(discount_rate), type(welcome_message))
print("is_active is bool:", isinstance(is_active, bool))
```

**Additional Example**: Mixing types in expressions.


```python
count = 3
label = 'items'
print(str(count) + ' ' + label)
print(isinstance(3.0, float), isinstance(3, int))
```

## 2. Literals and Numeral Systems

### Integer literals

- **Decimal**: `42`, `0`
- **Binary**: prefix `0b` or `0B` – e.g. `0b1010` → 10
- **Octal**: prefix `0o` or `0O` – e.g. `0o12` → 10
- **Hexadecimal**: prefix `0x` or `0X` – e.g. `0xFF` → 255

### Float literals

- Decimal: `3.14`, `.5`, `2.`
- **Scientific notation**: `1.5e3` (= 1500.0), `1e-2` (= 0.01)

### String literals

- Single or double quotes: `"hello"`, `'world'`
- Escape sequences: `\\`, `\n`, `\t`, `\"`, `\'`
- Raw strings: `r"C:\\path"` – backslashes not interpreted

```python
print("0b1010 =", 0b1010)
print("0o12 =", 0o12)
print("0xFF =", 0xFF)
print("1.5e3 =", 1.5e3)
print("1e-2 =", 1e-2)
```

**Additional Example**: Raw strings and escapes.


```python
path = r'C:\Users\data'
print('Raw:', path)
print('Tab:', 'a\tb')
print('Newline:', 'line1\nline2')
```

## 3. Type Conversion Built-ins

| Function | Purpose | Example |
|----------|--------|--------|
| **int(x)** | Convert to integer (from float truncates; from str parses) | `int("42")` → 42, `int(3.9)` → 3 |
| **float(x)** | Convert to float | `float("3.14")` → 3.14 |
| **str(x)** | String representation of x | `str(42)` → `"42"` |
| **bool(x)** | Boolean: False for 0, None, ""; True otherwise | `bool(1)` → True, `bool("")` → False |

### Optional base for int()

- **int(string, base)** – parse string in given base (2–36).
- Example: `int("FF", 16)` → 255, `int("1010", 2)` → 10.

```python
print(int("42"), int(3.9), int(True))
print(float("3.14"), float(42))
print(str(42), str(True))
print(bool(0), bool(1), bool(""), bool("hello"))
print("int('FF', 16) =", int("FF", 16))
```

**Additional Example**: Chained conversion.


```python
raw = '3.14'
print(float(raw), int(float(raw)))
print(bool('0'), bool(0))  # non-empty str is truthy
```

## 4. Arithmetic and Assignment Operators (PCEP 1.4)

### Arithmetic

| Operator | Meaning | Example |
|----------|--------|--------|
| `+` | Addition (or concatenation for str) | `3 + 5` → 8 |
| `-` | Subtraction | `10 - 4` → 6 |
| `*` | Multiplication (or repeat for str/list) | `3 * 4` → 12 |
| `/` | True division (result float) | `10 / 4` → 2.5 |
| `//` | Floor division | `10 // 4` → 2 |
| `%` | Remainder (modulo) | `10 % 4` → 2 |
| `**` | Exponentiation | `2 ** 3` → 8 |

### Assignment and augmented assignment

- `=` assigns to a name.
- `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, `**=` update in place: `x += 1` is equivalent to `x = x + 1` (for immutable types a new value is bound).

```python
print("10 / 4 =", 10 / 4)
print("10 // 4 =", 10 // 4)
print("10 % 4 =", 10 % 4)
print("2 ** 10 =", 2 ** 10)

x = 5
x += 3
print("x after x += 3:", x)
```

**Additional Example**: String repetition and floor division.


```python
print('ha' * 3)
print('17 // 5 =', 17 // 5, '17 % 5 =', 17 % 5)
```

## 5. Comparison and Boolean Operators

### Comparison (relational)

- `==` equal, `!=` not equal
- `<`, `<=`, `>`, `>=`
- Result is **bool** (`True` or `False`).

### Boolean operators

- **and** – True only if both operands are truthy. Short-circuits.
- **or** – True if at least one operand is truthy. Short-circuits.
- **not** – negates the boolean.

### Truthiness

- **Falsy**: `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `()`
- Everything else is **truthy**.

```python
print(3 == 3, 3 != 4, 2 < 5, 5 >= 5)
print(True and False, True or False, not True)

# Short-circuit: second part not evaluated if first determines result
# (False and ...) -> False; (True or ...) -> True
```

**Additional Example**: Chained comparisons and short-circuit.


```python
n = 15
print(10 <= n <= 20)  # chained comparison
print('and short-circuit:', False and print('skipped'))
```

## 6. Bitwise Operators (PCEP 1.4)

| Operator | Meaning | Example |
|----------|--------|--------|
| `~` | Bitwise NOT (invert bits) | `~0` → -1 |
| `&` | Bitwise AND | `5 & 3` → 1 |
| `|` | Bitwise OR | `5 | 3` → 7 |
| `^` | Bitwise XOR | `5 ^ 3` → 6 |
| `<<` | Left shift | `1 << 3` → 8 |
| `>>` | Right shift | `8 >> 2` → 2 |

Used for flags, masks, and low-level operations. Do not confuse `&`/`|` with `and`/`or`.

```python
print("5 & 3 =", 5 & 3)
print("5 | 3 =", 5 | 3)
print("1 << 4 =", 1 << 4)
print("16 >> 2 =", 16 >> 2)
```

**Additional Example**: Bit flags with AND/OR.


```python
READ, WRITE, EXEC = 1, 2, 4
perms = READ | WRITE
print('has READ:', bool(perms & READ))
print('has EXEC:', bool(perms & EXEC))
```

## 7. Operator Precedence and Binding

### Order (high to low, simplified)

1. Parentheses `()`
2. `**` (right-associative)
3. `+x`, `-x`, `~x` (unary)
4. `*`, `/`, `//`, `%`
5. `+`, `-`
6. `<<`, `>>`
7. `&`, `^`, `|`
8. Comparisons (`<`, `<=`, `>`, `>=`, `==`, `!=`)
9. `not`, `and`, `or`

**Best practice**: Use parentheses to make intent clear instead of relying on memory.

```python
print("1 + 2 * 3 =", 1 + 2 * 3)
print("(1 + 2) * 3 =", (1 + 2) * 3)
print("2 ** 3 ** 2 =", 2 ** 3 ** 2)  # right-associative: 2**(3**2)
```

**Additional Example**: Parentheses clarify precedence.


```python
print('not True and False:', not True and False)
print('not (True and False):', not (True and False))
```

## 8. Floating-Point and Numeric Built-ins

### Float accuracy

- Floats are approximate (binary representation).
- `0.1 + 0.2` is not exactly `0.3`. Compare with a **tolerance**, not `==`.

### Useful built-ins

| Function | Purpose | Example |
|----------|--------|--------|
| **abs(x)** | Absolute value | `abs(-5)` → 5 |
| **round(x, n)** | Round to n decimal places (n optional) | `round(3.14159, 2)` → 3.14 |
| **min(a, b, ...)** or **min(iterable)** | Minimum | `min(3, 1, 2)` → 1 |
| **max(a, b, ...)** or **max(iterable)** | Maximum | `max([1, 5, 2])` → 5 |
| **sum(iterable)** | Sum of items | `sum([1, 2, 3])` → 6 |
| **divmod(a, b)** | (a // b, a % b) | `divmod(10, 3)` → (3, 1) |
| **pow(base, exp)** | base ** exp | `pow(2, 3)` → 8 |

```python
print("0.1 + 0.2 =", 0.1 + 0.2)

def is_close(a: float, b: float, eps: float = 1e-9) -> bool:
    return abs(a - b) < eps

print("is_close(0.1+0.2, 0.3):", is_close(0.1 + 0.2, 0.3))
print("round(3.14159, 2):", round(3.14159, 2))
print("divmod(10, 3):", divmod(10, 3))
```

**Additional Example**: Bulk pricing with min/max.


```python
prices = [9.99, 14.50, 7.25, 22.00]
print('cheapest:', min(prices), 'total:', round(sum(prices), 2))
```

## 9. Console I/O: print() and input() (PCEP 1.5)

### print()

- **print(*objects, sep=' ', end='\\n', file=sys.stdout)**
- Converts each object to string and writes to `file` (default stdout).
- **sep** – string inserted between objects (default space).
- **end** – string appended after the last object (default newline).

### input()

- **input(prompt=None)** – writes `prompt` to stdout, reads one line from stdin, and returns it as a **string** (without the trailing newline).
- To get numbers, convert with **int()** or **float()** and use **try/except** for invalid input.

```python
print("A", "B", "C", sep="-", end="!\n")

# input() returns str; uncomment to try in notebook (may not work in all environments):
# raw = input("Enter a number: ")
# num = int(raw)  # ValueError if not a valid integer
```
