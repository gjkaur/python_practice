## 1. String Creation and Immutability

- Single or double quotes: `"hello"`, `'world'`.
- Strings are **immutable**: no in-place change. Methods return **new** strings.
- **str(x)** – built-in to convert any object to its string representation.

```python
s = "Hello"
# s[0] = "h"  # TypeError
upper = s.upper()
print("original:", s, "after .upper():", upper)
print("str(42):", str(42))
```

**Additional Example**: String indexing and slicing.


```python
s = 'Python'
print(s[0], s[-1], s[2:5])
```

---

## 2. Indexing and Slicing

- **s[i]** – character at index i (0-based). Negative: -1 is last character.
- **s[start:stop]** – substring from start to stop-1. Omitted start/stop default to 0/len(s). Slicing does not raise IndexError.
- **s[start:stop:step]** – with step (e.g. s[::-1] reverses the string).

```python
name = "Alice"
print("name[0]:", name[0])
print("name[-1]:", name[-1])
print("name[1:4]:", name[1:4])
print("name[:3]:", name[:3])
print("name[::-1]:", name[::-1])
```

**Additional Example**: Case conversion methods.


```python
msg = 'Hello World'
print(msg.lower(), msg.upper(), msg.title())
```

## 3. Escape Sequences and Multi-line Strings

- **\\** – backslash. **\\n** – newline. **\\t** – tab. **\\"** – double quote. **\\'** – single quote.
- **Raw string**: r"..." – backslashes not interpreted (e.g. r"C:\\path").
- **Triple-quoted** """...""" or '''...''' – multi-line string; newlines preserved.

```python
print("He said, \"Hi\"")
print("Line1\nLine2")
ml = """First line
Second line"""
print(ml)
print(repr(ml))
```

**Additional Example**: split and join.


```python
csv = 'apple,banana,cherry'
fruits = csv.split(',')
print('split:', fruits)
print('join:', ' | '.join(fruits))
```

## 4. Case and Whitespace Methods

| Method | Description | Example |
|--------|-------------|--------|
| **lower()** | All lowercase | "Hello".lower() -> "hello" |
| **upper()** | All uppercase | "Hello".upper() -> "HELLO" |
| **title()** | Title case (each word capitalized) | "hi there".title() -> "Hi There" |
| **capitalize()** | First char upper, rest lower | "hello".capitalize() -> "Hello" |
| **strip()** | Remove leading/trailing whitespace | "  a  ".strip() -> "a" |
| **lstrip()**, **rstrip()** | Strip left or right only | |

Use **strip()** and **lower()** to normalize user input before comparison.

```python
print("  yes  ".strip())
print("Hello World".lower())
print("hello world".title())
raw = "  abc  "
print("|", raw.strip(), "|")
```

**Additional Example**: strip whitespace.


```python
raw = '  hello  \n'
print(repr(raw.strip()))
print(repr(raw.lstrip()))
```

## 5. Search Methods: find, index, count, startswith, endswith

| Method | Description | Returns |
|--------|-------------|--------|
| **find(sub)** | Index of first sub; -1 if not found | int |
| **rfind(sub)** | Index of last sub; -1 if not found | int |
| **index(sub)** | Like find but **ValueError** if not found | int |
| **count(sub)** | Number of non-overlapping occurrences | int |
| **startswith(prefix)** | True if string starts with prefix | bool |
| **endswith(suffix)** | True if string ends with suffix | bool |

**in** operator: **sub in s** – True if sub is a substring.

```python
s = "hello world"
print("find('o'):", s.find("o"))
print("rfind('o'):", s.rfind("o"))
print("count('l'):", s.count("l"))
print("startswith('hel'):", s.startswith("hel"))
print("'world' in s:", "world" in s)
```

**Additional Example**: find and replace.


```python
text = 'one two one'
print('find one:', text.find('one'))
print('replace:', text.replace('one', '1'))
```

## 6. replace, split, join

| Method | Description | Example |
|--------|-------------|--------|
| **replace(old, new)** | Replace all occurrences of old with new | "a-b".replace("-", " ") |
| **split(sep=None)** | Split into list by sep (default: whitespace) | "a b c".split() -> ["a","b","c"] |
| **rsplit(sep=None)** | Split from the right (maxsplit optional) | |
| **splitlines()** | Split by line boundaries | "a\\nb".splitlines() -> ["a","b"] |
| **sep.join(iterable)** | Join strings in iterable with sep (str method) | ", ".join(["a","b"]) -> "a, b" |

**Note**: **join** is called on the *separator* string: `" ".join(words)`.

```python
line = "alice,bob,charlie"
names = line.split(",")
print("split(','):", names)
joined = ", ".join(names)
print("join:", joined)
print("replace:", "hello".replace("l", "L"))
```

**Additional Example**: f-string formatting.


```python
item, price = 'book', 19.99
print(f'{item}: ${price:.2f}')
```

## 7. Formatting: f-strings and format()

- **f"...{expr}..."** – f-string: expressions in braces are evaluated and formatted.
- **"...{}".format(...)** – older style; `{}` placeholders filled by format() arguments.
- Format spec: **{x:.2f}** (2 decimals), **{x:5d}** (width 5), etc.

Prefer **f-strings** for readability.

```python
name = "Alice"
score = 85.5
print(f"{name} scored {score}.")
print(f"Score: {score:.2f}")
print("{} scored {}.".format(name, score))
```

**Additional Example**: Character membership.


```python
word = 'code'
print('c' in word, 'z' in word)
print(''.join(reversed(word)))
```

## 8. Predicate Methods: isalpha, isdigit, isalnum, isspace

| Method | True when |
|--------|----------|
| **isalpha()** | All characters are letters (and at least one) |
| **isdigit()** | All characters are digits (and at least one) |
| **isalnum()** | All characters are alphanumeric |
| **isspace()** | All characters are whitespace (and at least one) |
| **isupper()**, **islower()** | All cased characters are upper/lower |

Empty string returns **False** for these.

```python
print("abc".isalpha(), "123".isdigit(), "a1".isalnum())
print("  \t".isspace(), "".isspace())
```

**Additional Example**: startswith / endswith.


```python
filename = 'report_2024.txt'
print(filename.startswith('report'), filename.endswith('.txt'))
```

## 9. Built-ins: str(), repr(), len()

- **str(x)** – human-readable string representation of x.
- **repr(x)** – developer-oriented representation (often with quotes for strings).
- **len(s)** – number of characters in string s.

```python
print("str(42):", str(42))
print("repr('hi'):", repr("hi"))
print("len('hello'):", len("hello"))
```
