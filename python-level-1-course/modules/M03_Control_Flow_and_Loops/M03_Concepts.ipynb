## 1. Boolean Expressions and Short-Circuit

### and / or / not

- **and**: True only if both operands are truthy. Stops at first falsy (short-circuit).
- **or**: True if any operand is truthy. Stops at first truthy (short-circuit).
- **not**: Returns the boolean opposite.

### Use parentheses to clarify

- `(a or b) and c` is clearer than relying on precedence.

### Truthiness (recap)

- Falsy: `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `()`
- Everything else is truthy.

````python
is_premium = True
cart_total = 120.0
eligible = (is_premium or cart_total >= 100) and cart_total > 0
print("Eligible for discount:", eligible)

# Short-circuit: if first is False, second is not evaluated
print(False and (1/0))  # no error: 1/0 not evaluated
````

**Additional Example**: if / elif / else chain.


````python
score = 85
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
else:
    grade = 'C'
print(f'Score {score} -> {grade}')
````

## 2. if / elif / else – Syntax and Patterns

### Syntax

```python
if condition1:
    block1
elif condition2:
    block2
else:
    block3
```

- **elif** and **else** are optional.
- At most one branch runs; conditions are checked in order.

### Guard clauses

- Return or exit early for invalid/special cases to avoid deep nesting.
- Example: `if not is_valid: return False` then main logic.

````python
def classify_score(score: int) -> str:
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

print(classify_score(85))

# Guard clause style
def can_access(is_active: bool, is_admin: bool) -> bool:
    if not is_active:
        return False
    return is_admin
````

**Additional Example**: Nested conditionals.


````python
age, has_ticket = 16, True
if age >= 18:
    print('Adult entry')
elif has_ticket and age >= 13:
    print('Teen with ticket')
else:
    print('No entry')
````

## 4. for Loops and range()

### for loop syntax

```python
for item in iterable:
    block
```

### range() – built-in

- **range(stop)** – integers 0, 1, ..., stop-1
- **range(start, stop)** – start, start+1, ..., stop-1
- **range(start, stop, step)** – step can be negative
- **range** is lazy (does not build a list in memory).

### Converting to list

- **list(range(5))** → `[0, 1, 2, 3, 4]`

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 6):
    print(i, end=" ")
print()

for i in range(10, 0, -2):
    print(i, end=" ")
print()

print("list(range(3)):", list(range(3)))
````

**Additional Example**: for loop over a string.


````python
word = 'Python'
for ch in word:
    print(ch, end=' ')
print()
````

## 5. The in Operator

- **x in s** – True if x is a member of s (element of sequence or key in dict).
- **x not in s** – negation.
- Used in **for item in sequence** and in **if item in sequence**.

````python
print("3 in [1,2,3]:", 3 in [1, 2, 3])
print("10 not in range(5):", 10 not in range(5))
print("'a' in 'hello':", "a" in "hello")

for letter in "abc":
    print(letter, end=" ")
````

**Additional Example**: break exits early.


````python
for i in range(10):
    if i == 5:
        break
    print(i, end=' ')
print('(stopped at 5)')
````

## 6. while Loops

### Syntax

```python
while condition:
    block
```

- **condition** is evaluated before each iteration; if False, the loop exits.
- Ensure the condition can become False (or use **break**), or you get an infinite loop.

### Common pattern: menu loop

```python
while True:
    choice = input("Option: ")
    if choice == "quit":
        break
    # handle choice
```

````python
n = 0
while n < 3:
    print("n =", n)
    n += 1
````

**Additional Example**: continue skips iteration.


````python
for n in range(1, 6):
    if n % 2 == 0:
        continue
    print(n, end=' ')
print('(odds only)')
````

## 7. break and continue

- **break** – exits the innermost **for** or **while** loop immediately.
- **continue** – skips the rest of the current iteration and goes to the next one.

Both only affect the loop they are in (not outer loops).

````python
# break: exit on 3
for i in range(5):
    if i == 3:
        break
    print(i, end=" ")
print(" (after loop)")

# continue: skip 2
for i in range(4):
    if i == 2:
        continue
    print(i, end=" ")
print(" (after loop)")
````

**Additional Example**: else on a for-loop (no break).


````python
for x in [2, 4, 6]:
    if x % 2:
        break
else:
    print('All numbers were even')
````

## 8. for-else and while-else

- The **else** clause of a **for** or **while** runs only if the loop **completed normally** (did not exit via **break**).
- Useful for search loops: "if we broke out we found it; else we didn't find it."

````python
def find_first_over(values: list, limit: int):
    for v in values:
        if v > limit:
            return v
    else:
        return None  # no break: nothing found

print(find_first_over([1, 2, 5, 3], 4))
print(find_first_over([1, 2], 10))
````

**Additional Example**: Simple menu loop pattern.


````python
choice = '2'  # simulate user picking option 2
if choice == '1':
    print('View items')
elif choice == '2':
    print('Add item')
else:
    print('Invalid choice')
````

## 9. Nested Loops and Conditionals

- You can put **if** inside **for**, **for** inside **for**, **while** inside **if**, etc.
- Keep nesting shallow; use guard clauses or helper functions to avoid "pyramid of doom."

### enumerate()

- **enumerate(iterable, start=0)** – yields (index, item) pairs. Useful when you need the index in a for loop.

````python
# Nested: print pairs (i, j) where i < j
for i in range(3):
    for j in range(3):
        if i < j:
            print((i, j), end=" ")
print()

# enumerate
for idx, letter in enumerate("ab", start=1):
    print(idx, letter)
````

**Practice 6:** Write a loop that finds the first index where a list element is negative. Use break. If none, print "No negative".

````python
data = [1, 3, -2, 5]
idx = None
for i, x in enumerate(data):
    if x < 0:
        idx = i
        break
if idx is not None:
    print("First negative at index:", idx)
else:
    print("No negative")
````
