## 1. Sequences and Common Built-ins

### len(s)

- **len(s)** – returns number of items in a sequence or mapping.
- Lists, tuples, strings, dicts (number of keys) all support **len()**.

### sorted(iterable)

- **sorted(iterable)** – returns a **new list** of all items in ascending order.
- Does not modify the original. For lists, **.sort()** sorts in place.

### del statement

- **del obj** or **del sequence[index]** or **del dict[key]** – deletes a name or an item.
- Example: `del my_list[0]`; `del d["key"]`.

```python
print("len([1,2,3]):", len([1, 2, 3]))
print("len('hello'):", len("hello"))
print("sorted([3,1,2]):", sorted([3, 1, 2]))

lst = [10, 5, 20]
del lst[1]
print("After del lst[1]:", lst)
```

## 2. Lists – Creation, Indexing, Slicing (PCEP 3.1)

### Creation

- **[]** – empty list.
- **[a, b, c]** – list literal.
- **list(iterable)** – build list from iterable (e.g. list(range(5)), list("abc")).

### Indexing

- **lst[i]** – element at index i (0-based). Negative index: -1 is last, -2 second-to-last.
- **IndexError** if index out of range.

### Slicing

- **lst[start:stop]** – elements from start to stop-1.
- **lst[start:stop:step]** – with step. Omitted start/default 0; omitted stop/default len. Slicing never raises; returns empty or shorter list.

```python
students = ["Alice", "Bob", "Charlie", "Diana"]
print("students[0]:", students[0])
print("students[-1]:", students[-1])
print("students[1:3]:", students[1:3])
print("students[:2]:", students[:2])
print("students[-2:]:", students[-2:])
print("students[::2]:", students[::2])
```

## 3. List Methods (PCEP 3.1)

### Mutating (change the list in place)

| Method | Description | Example |
|--------|-------------|--------|
| **append(x)** | Add x at end | lst.append(4) |
| **insert(i, x)** | Insert x at index i | lst.insert(0, 0) |
| **extend(iterable)** | Append all items | lst.extend([5,6]) |
| **remove(x)** | Remove first occurrence of x; ValueError if not found | lst.remove(2) |
| **pop([i])** | Remove and return item at i (default last) | lst.pop() |
| **clear()** | Remove all items | lst.clear() |
| **sort()** | Sort in place (optional key, reverse) | lst.sort() |
| **reverse()** | Reverse in place | lst.reverse() |

### Non-mutating (return value, list unchanged)

| Method | Description | Example |
|--------|-------------|--------|
| **index(x)** | Index of first x; ValueError if not found | lst.index(3) |
| **count(x)** | Number of occurrences of x | lst.count(2) |
| **copy()** | Shallow copy (new list) | lst.copy() |

```python
lst = [1, 2, 3]
lst.append(4)
print("After append(4):", lst)
lst.insert(0, 0)
print("After insert(0, 0):", lst)
print("index(2):", lst.index(2))
print("count(2):", lst.count(2))

lst2 = lst.copy()
lst2.pop()
print("original:", lst, "copy after pop:", lst2)
```

## 4. List Comprehensions

- **[expr for item in iterable]** – new list of expr for each item.
- **[expr for item in iterable if condition]** – filter with if.
- Can use nested loops: **[x+y for x in a for y in b]**.

Use for readable, compact construction of lists.

```python
scores = [65, 80, 90, 72]
passing = [s for s in scores if s >= 70]
print("passing:", passing)

squared = [x**2 for x in range(5)]
print("squared:", squared)

# With else-like logic: use conditional expression
labels = ["pass" if s >= 70 else "fail" for s in scores]
print("labels:", labels)
```

## 5. Copy vs Alias

- **b = a** – b and a refer to the **same** list (alias). Mutating through one affects the other.
- **b = a.copy()** or **b = list(a)** – **shallow** copy: new list, but elements are the same objects.
- For lists of immutables (numbers, strings), shallow copy is usually enough.
- **copy.deepcopy()** (from module copy) for nested structures if you need full copy.

```python
a = [1, 2, 3]
b = a
b.append(4)
print("a:", a)  # a changed!

c = a.copy()
c.append(5)
print("a:", a, "c:", c)
```

## 6. Tuples – Immutable Sequences (PCEP 3.2)

### Creation

- **()** – empty tuple (rare).
- **(a,)** – single-element tuple (comma required).
- **(a, b, c)** or **a, b, c** – tuple of three.
- **tuple(iterable)** – build from iterable.

### Why tuples

- **Immutable** – cannot add, remove, or change elements. Safe as dict keys or set elements.
- Fixed-length "records": (name, age), (x, y).

### Tuple methods (only two)

- **index(x)** – index of first x.
- **count(x)** – count of x.

```python
user = ("alice", "alice@example.com")
print(user[0], user[1])

# Unpacking
name, email = user
print(name, email)

# Single-element tuple
t = (42,)
print(type(t), t)

# tuple() from range
print(tuple(range(3)))
```

## 7. Dictionaries – Key-Value Mapping (PCEP 3.3)

### Creation

- **{}** – empty dict.
- **{k1: v1, k2: v2}** – literal.
- **dict()** – empty or from keyword args: dict(a=1, b=2).
- **dict([(k,v), ...])** – from list of pairs.

### Keys

- Keys must be **hashable** (immutable: str, int, float, tuple of hashables). No lists as keys.
- **d[key]** – get value; **KeyError** if key missing.
- **d.get(key, default)** – get value or default if key missing (no error).
- **key in d** – membership test.

### Mutating

- **d[key] = value** – set or update.
- **del d[key]** – remove key; KeyError if missing.

```python
inventory = {"SKU001": 10, "SKU002": 3}
inventory["SKU003"] = 0
print("inventory:", inventory)
print("get('SKU999', 0):", inventory.get("SKU999", 0))
print("'SKU001' in inventory:", "SKU001" in inventory)
```

## 8. Dict Methods and Iteration

### Methods

| Method | Description |
|--------|-------------|
| **keys()** | View of keys |
| **values()** | View of values |
| **items()** | View of (key, value) pairs |
| **get(k, default)** | Value for k or default |
| **setdefault(k, default)** | Get k; if missing set d[k]=default and return default |
| **update(other)** | Merge other dict (or iterable of pairs) into d |
| **pop(k)** | Remove k and return its value; KeyError if missing |
| **popitem()** | Remove and return one (key, value) pair |
| **clear()** | Remove all |

### Iteration

- **for k in d:** iterates keys.
- **for k, v in d.items():** iterates key-value pairs.

```python
d = {"a": 1, "b": 2, "c": 3}
print("keys:", list(d.keys()))
print("values:", list(d.values()))
for k, v in d.items():
    print(k, v)

d.setdefault("d", 4)
print("After setdefault('d', 4):", d)
print("pop('b'):", d.pop("b"), "-> d:", d)
```

## 9. Built-ins: len, sorted, del (recap)

- **len(d)** – number of key-value pairs.
- **sorted(d)** – sorted list of keys (or use key= for custom order).
- **del d[k]** – remove key k.

### list(), tuple(), dict()

- **list(iterable)** – new list from iterable.
- **tuple(iterable)** – new tuple.
- **dict()** or **dict(mapping)** or **dict(iterable of pairs)** – new dict.

```python
d = {"z": 3, "a": 1, "m": 2}
print("sorted(d):", sorted(d))
print("sorted(d.keys(), key=lambda k: d[k]):", sorted(d.keys(), key=lambda k: d[k]))
```

## More Examples

**Example: List methods – insert, remove, reverse**

```python
lst = [1, 2, 4]
lst.insert(2, 3)   # insert 3 at index 2
print(lst)
lst.remove(2)      # remove first 2
print(lst)
lst.reverse()
print("Reversed:", lst)
```

**Example: Dict update and setdefault**

```python
d = {"a": 1, "b": 2}
d.update({"b": 20, "c": 3})  # merge; b overwritten
print(d)
print("setdefault('d', 4):", d.setdefault("d", 4))
print("setdefault('a', 99):", d.setdefault("a", 99))  # a exists, returns 1
print(d)
```
