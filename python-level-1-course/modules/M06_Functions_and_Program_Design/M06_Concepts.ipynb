## 2. return and None

- **return expression** – exits the function and returns the value.
- **return** with no expression, or no return at all – returns **None**.
- **None** is the single value of type **NoneType**. Use **-> None** in type hints for procedures.

````python
def calculate_tax(amount: float, rate: float) -> float:
    return amount * rate

print(calculate_tax(100.0, 0.2))
````

**Additional Example**: Implicit None return.


````python
def log(msg):
    print('[LOG]', msg)

result = log('started')
print('Return value:', result)
````

## 3. Parameters vs Arguments; Positional and Keyword

- **Parameter** – name in the function definition. **Argument** – value passed when calling.
- **Positional** arguments matched by order. **Keyword** arguments: name=value; order does not matter.
- Mixed: positional must come before keyword.

**Additional Example**: Positional vs keyword args.


````python
def connect(host, port=8080, ssl=False):
    return f'{host}:{port} ssl={ssl}'

print(connect('localhost'))
print(connect(port=443, host='api.example.com', ssl=True))
````

## 2. Default arguments

- Defaults allow optional parameters.
- **Avoid mutable defaults**: `def f(x, acc=[])` is dangerous (shared list).

````python
def greet(name: str, greeting: str) -> str:
    return f"{greeting}, {name}!"
print(greet("Alice", "Hello"))
print(greet(greeting="Hi", name="Bob"))
````

````python
def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}!"

print(greet("Alice"))
print(greet("Bob", greeting="Hi"))
````

**Additional Example**: Implicit None return.


````python
def log(msg):
    print('[LOG]', msg)

result = log('started')
print('Return value:', result)
````

## 4. Scope: Local, Global, and global keyword

- **Local**: names assigned inside the function (parameters and variables).
- **Global**: names at module level. **LEGB** lookup: Local, Enclosing, Global, Built-in.
- To **rebind** a global name inside a function, use **global name**. Prefer passing and returning instead.

**Additional Example**: Local vs global (read-only).


````python
tax_rate = 0.2

def add_tax(amount):
    return amount * (1 + tax_rate)  # reads global, no global keyword needed

print(add_tax(100))
````

## 3. Scope

- Local names inside function; global at module level.
- Use `global name` sparingly; prefer passing values and returning.

**Additional Example**: Positional vs keyword args.


````python
def connect(host, port=8080, ssl=False):
    return f'{host}:{port} ssl={ssl}'

print(connect('localhost'))
print(connect(port=443, host='api.example.com', ssl=True))
````

## 5. Docstrings and help()

- **Docstring**: string literal as the first statement in a function; use triple quotes.
- **help(function)** prints the docstring. **function.__doc__** gives the string.

````python
def increment(value: int) -> int:
    return value + 1  # no global needed

x = 5
x = increment(x)
print(x)
````

````python
def add(a: int, b: int) -> int:
    """Return the sum of a and b."""
    return a + b
print(add.__doc__)
print("callable(add):", callable(add))
````

**Additional Example**: Docstring and help.


````python
def median(values):
    """Return middle value of a sorted list (simple odd-length case)."""
    s = sorted(values)
    return s[len(s) // 2]

print(median.__doc__)
print(median([3, 1, 2]))
````

## 4. None and return

- Functions without a return (or `return` alone) return `None`.
- Type hints: `-> None` for procedures.

````python
def log_message(msg: str) -> None:
    print("[LOG]", msg)
    # no return

result = log_message("test")
print(result is None)
````

**Additional Example**: Local vs global (read-only).


````python
tax_rate = 0.2

def add_tax(amount):
    return amount * (1 + tax_rate)  # reads global, no global keyword needed

print(add_tax(100))
````

## More Examples

**Example: Function that returns None for "not found"**

````python
def find_index(lst, target):
    """Return index of target in lst, or None if not found."""
    for i, x in enumerate(lst):
        if x == target:
            return i
    return None

print(find_index([10, 20, 30], 20))
print(find_index([10, 20, 30], 99))
````

**Example: Default argument with None (safe mutable default)**

````python
def add_to_list(item, container=None):
    if container is None:
        container = []
    container.append(item)
    return container

print(add_to_list(1))   # [1]
print(add_to_list(2))   # [2] – new list each time, not [1, 2]
````

**Practice 6:** Write a function that takes a list and returns (min, max). Return (None, None) for an empty list.

````python
def min_max(lst):
    if not lst:
        return (None, None)
    return (min(lst), max(lst))
print(min_max([3, 1, 4, 2]))
print(min_max([]))
````
