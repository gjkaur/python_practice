## 1. Interpreter vs Compiler (PCEP 1.1)

### Concepts

- **Compiler**: Translates source code into a separate executable *before* running (e.g. C, Go).
- **Interpreter**: Reads and executes source (or bytecode) step by step. Python is typically run by the **CPython interpreter**.

### How Python Runs

- **REPL** (Read-Eval-Print Loop): Type `python` or `python3` in a terminal; enter statements one at a time.
- **Script**: Save code in a `.py` file; run with `python script.py`.
- Internally, CPython compiles to bytecode (`.pyc`) for speed, but we still speak of "interpreted" execution.

### Related

- `sys.version` – Python version string.
- `sys.executable` – Path to the Python interpreter binary.

````python
import sys
print("Python version:", sys.version)
print("Executable:", sys.executable)
````

**Additional Example**: Compare script vs interactive execution.


````python
print('Running in notebook — same as a .py script line by line')
print(2 + 2)
````

## 2. Lexis, Syntax, and Semantics

### Definitions

- **Lexis**: The "vocabulary" of the language – keywords, identifiers, literals (numbers, strings), operators, punctuation.
- **Syntax**: Rules that define *grammatically valid* combinations of tokens (e.g. an `if` must be followed by `:` and an indented block).
- **Semantics**: The *meaning* of valid code – what actually happens when it runs.

### Example Breakdown

```python
message = "Hello"
print(message)
```

- **Lexis**: `message` (identifier), `=` (operator), `"Hello"` (string literal), `print` (identifier), `(`, `)`, `message`.
- **Syntax**: Assignment statement; function call expression statement.
- **Semantics**: Bind the name `message` to the string `"Hello"`; then call `print` with that value and output it to stdout.

### Syntax vs Runtime Errors

- **SyntaxError**: Invalid grammar (e.g. missing `:`, wrong indent). Caught when the module is loaded.
- **Runtime errors**: Code is syntactically valid but fails during execution (e.g. `NameError`, `ZeroDivisionError`).

````python
# Lexis: identifiers, literals, operators
message = "Hello"
print(message)

# This would be SyntaxError in Python 3 (uncomment to see):
# print "Hello"
````

**Additional Example**: Syntax vs semantics — valid syntax, runtime semantics.


````python
# Syntax OK; semantics: division by zero raises ZeroDivisionError
try:
    result = 1 / 0
except ZeroDivisionError as e:
    print('Semantic error caught:', type(e).__name__)
````

## 3. Python Keywords and Identifiers

### Keywords

**Keywords** are reserved words that have special meaning. You cannot use them as variable or function names.

**Built-in function** `keyword.kwlist` lists all keywords. Common ones you will use early: `if`, `else`, `elif`, `for`, `while`, `def`, `return`, `True`, `False`, `None`, `and`, `or`, `not`, `in`, `is`, `lambda`, `with`, `try`, `except`, `finally`, `raise`, `import`, `from`, `as`, `class`, `pass`, `break`, `continue`.

### Identifiers

- Names for variables, functions, classes, modules.
- Rules: Start with a letter or `_`; rest can be letters, digits, `_`.
- PEP 8: use `snake_case` for variables and functions; `UPPER_SNAKE_CASE` for constants.

````python
import keyword
print("Python keywords:")
print(keyword.kwlist)
print("\nIs 'for' a keyword?", keyword.iskeyword("for"))
print("Is 'count' a keyword?", keyword.iskeyword("count"))
````

**Additional Example**: Valid identifiers and keyword check.


````python
import keyword
names = ['total_count', '_private', 'score2', '2bad']
for name in names:
    valid = name.isidentifier() and not keyword.iskeyword(name)
    print(f'{name!r}: valid={valid}')
````

## 4. Indentation and Instructions (PCEP 1.2)

### Rules

- Python uses **indentation** to define blocks (no braces `{}`).
- **PEP 8**: Use **4 spaces** per indentation level. Do not mix tabs and spaces.
- The same block must use the same indent consistently.
- **IndentationError** occurs if indentation is inconsistent or wrong.

### What Counts as a Block

- Body of `if`, `elif`, `else`.
- Body of `for`, `while`.
- Body of `def`, `class`.
- Body of `try`, `except`, `finally`, `with`.

````python
value = 10

if value > 5:
    print("Greater than five")   # 4 spaces
    print("Still in block")     # same 4 spaces

# Next line at column 0: outside the if block
print("Done")
````

**Additional Example**: Indentation defines blocks.


````python
x = 10
if x > 5:
    print('Indented block runs')
    print('Still inside if')
print('Back at top level')
````

## 5. Comments and Docstrings

### Single-line comments

- Everything after `#` on that line is ignored by the interpreter.
- Use to explain *why*, not to restate the code.

### Docstrings

- A string literal as the first statement in a function, class, or module is a **docstring**.
- Used by `help()` and tools to document the object.
- Convention: triple-quoted string, e.g. `"""Describe what this does."""`

````python
# This is a comment. It is ignored when the code runs.

def greet(name: str) -> str:
    """Return a greeting string for the given name."""
    return f"Hello, {name}!"

# View the docstring:
print(greet.__doc__)
# Or: help(greet)
````

**Additional Example**: Single-line and block comments.


````python
# This is a comment
value = 42  # inline comment
print(value)  # output: 42
````

## 6. Script Structure and `__main__`

### `if __name__ == "__main__":`

- When you run a file with `python script.py`, Python sets the special variable `__name__` in that file to the string `"__main__"`.
- When the same file is **imported** as a module, `__name__` is set to the module name (e.g. `"mymodule"`).
- So code under `if __name__ == "__main__":` runs **only** when the file is executed directly, not when it is imported.

### Why Use It

- Keeps "entrypoint" code (e.g. calling `main()`) separate.
- Allows the file to be both run as a script and imported for its functions/classes without running the script part.

````python
def main() -> None:
    print("Hello from main()")

# This block runs only when this file is executed, not when imported
if __name__ == "__main__":
    main()

print("__name__ in this notebook is:", __name__)
````

**Additional Example**: `__name__` guard pattern.


````python
def main():
    print('App started')

print('__name__ is:', __name__)
if __name__ == '__main__':
    main()
````

## 7. PEP 8 and Naming Conventions

### Summary

| Kind | Convention | Example |
|------|------------|--------|
| Variables, functions | `snake_case` | `user_name`, `get_score()` |
| Constants | `UPPER_SNAKE_CASE` | `MAX_RETRIES`, `DEFAULT_PORT` |
| Files/modules | `lowercase_with_underscores` | `my_module.py` |
| Classes | `PascalCase` | `UserProfile` |
| Private (by convention) | leading `_` | `_internal_helper()` |

### Other PEP 8 Points

- Maximum line length 88–99 characters (tools like Black use 88).
- Two blank lines before top-level function/class definitions.
- One blank line between method definitions inside a class.
- Spaces around `=` in keyword arguments: `func(a=1)` not `func(a=1)` (no space before `=` in default values).

````python
MAX_RETRIES = 3
user_name = "Alice"

def get_greeting(name: str) -> str:
    return f"Hello, {name}!"

print(get_greeting(user_name))
````

## 8. Built-in Functions Used from Day One

These are always available (no import).

| Function | Purpose | Example |
|----------|--------|--------|
| `print(*objects, sep=' ', end='\n')` | Output to stdout | `print(1, 2, sep=',')` |
| `len(s)` | Number of items in a sequence or mapping | `len([1,2,3])` → 3 |
| `type(x)` | Type of object | `type(42)` → `<class 'int'>` |
| `id(x)` | Identity (memory address) of object | `id([])` |
| `help(x)` | Show docstring / help | `help(print)` |
| `input(prompt)` | Read a line from stdin (returns str) | `input("Name: ")` |
| `int(x)`, `float(x)`, `str(x)`, `bool(x)` | Type conversion | `int("42")` → 42 |

````python
# print() with sep and end
print("A", "B", "C", sep=" | ", end="!\n")

# len() works on sequences and dicts
print("len([10, 20]):", len([10, 20]))
print("len('hello'):", len("hello"))

# type() and id()
a = [1, 2]
print("type(a):", type(a))
print("id(a):", id(a))
````

**Additional Example**: Built-ins `len`, `type`, `print`.


````python
items = [10, 20, 30]
print('len:', len(items))
print('type:', type(items))
print('sum via built-in:', sum(items))
````

## 9. Environment and Module Introspection

### Useful attributes and modules

## 10. Common Errors and Tracebacks

### Error types you will see early

| Error | Typical cause |
|-------|----------------|
| **SyntaxError** | Invalid grammar (e.g. missing `:`, bad indent) |
| **IndentationError** | Inconsistent or wrong indentation |
| **NameError** | Name not defined (typo, or used before assignment) |
| **TypeError** | Wrong type (e.g. `int + str`) |
| **ValueError** | Valid type but invalid value (e.g. `int("abc")`) |

### How to read a traceback

1. **Last line**: The exception type and message.
2. **Above it**: The call stack – each "File ... line X, in function" shows where the call came from.
3. **Top**: Where the error was **raised**; bottom is where it was **first called**.

### Good habits

- Read the **last line** first for the error type and message.
- Then look at the **first file and line** that is *your* code (not library code) to fix the cause.
