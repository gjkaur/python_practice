## 1. Exception Chaining (PCPP 1.9)

Exception chaining links exceptions to show error propagation.

```python
class DatabaseError(Exception):
    pass

class ApplicationError(Exception):
    pass

def database_operation():
    raise DatabaseError("Connection timeout")

def save_user_data():
    try:
        database_operation()
    except DatabaseError as db_error:
        # Explicit chaining: raise NewException from OriginalException
        raise ApplicationError("Failed to save user data") from db_error

try:
    save_user_data()
except ApplicationError as e:
    print(f"Caught: {e}")
    print(f"Caused by: {e.__cause__}")
    print(f"Context: {e.__context__}")
```

**Additional Example**: Implicit vs explicit chaining.


```python
def low():
    raise ValueError('low level')

def high():
    try:
        low()
    except ValueError as e:
        raise RuntimeError('high level') from e

try:
    high()
except RuntimeError as e:
    print('cause type:', type(e.__cause__).__name__)
```

## 2. Traceback Objects (PCPP 1.9)

Access and analyze exception tracebacks programmatically.

```python
import traceback

try:
    raise ValueError("Test error")
except ValueError as e:
    # Access __traceback__ attribute
    tb = e.__traceback__
    
    # Format traceback
    formatted = traceback.format_exception(type(e), e, tb)
    print(''.join(formatted))
```

**Additional Example**: traceback.format_exc in except block.


```python
import traceback

def risky():
    return 1 / 0

try:
    risky()
except ZeroDivisionError:
    print(traceback.format_exc().splitlines()[-1])
```

## 3. Object Copying (PCPP 1.10)

Shallow copy vs deep copy: understanding when nested objects are shared.

```python
import copy

class NestedObject:
    def __init__(self, name: str, items: list):
        self.name = name
        self.items = items

original = NestedObject("Original", ["item1", "item2"])

# Shallow copy - nested objects are shared
shallow = copy.copy(original)
shallow.items.append("item3")
print(f"Original: {original.items}")  # Modified!
print(f"Shallow: {shallow.items}")

# Deep copy - completely independent
original2 = NestedObject("Original2", ["item1", "item2"])
deep = copy.deepcopy(original2)
deep.items.append("item3")
print(f"\nOriginal2: {original2.items}")  # Not modified
print(f"Deep: {deep.items}")
```

**Additional Example**: deepcopy nested dict.


```python
import copy
original = {'a': [1, 2], 'b': {'x': 1}}
shallow = copy.copy(original)
deep = copy.deepcopy(original)
shallow['a'].append(99)
deep['b']['x'] = 42
print('original:', original)
print('deep b:', deep['b'])
```
