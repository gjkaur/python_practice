## 1. Basic Window (PCPP 3.2)

Create a window with basic widgets.

```python
import tkinter as tk

root = tk.Tk()
root.title("My Window")
root.geometry("300x200")

label = tk.Label(root, text="Hello, tkinter!")
label.pack(pady=10)

button = tk.Button(root, text="Click Me", command=lambda: print("Clicked!"))
button.pack(pady=5)

# root.mainloop()  # Uncomment to run
```

## 2. Grid Layout (PCPP 3.2)

Table-like layout using grid manager.

```python
root = tk.Tk()
root.title("Grid Layout")

tk.Label(root, text="Name:").grid(row=0, column=0, padx=5, pady=5)
tk.Entry(root, width=20).grid(row=0, column=1, padx=5, pady=5)

tk.Label(root, text="Email:").grid(row=1, column=0, padx=5, pady=5)
tk.Entry(root, width=20).grid(row=1, column=1, padx=5, pady=5)

tk.Button(root, text="Submit").grid(row=2, column=0, columnspan=2, pady=10)

# root.mainloop()  # Uncomment to run
```

## 3. Event Handling (PCPP 3.3)

Handle user events with callbacks.

```python
def on_button_click():
    print("Button clicked!")

def on_key_press(event):
    print(f"Key: {event.char}")

root = tk.Tk()
button = tk.Button(root, text="Click", command=on_button_click)
button.pack()

entry = tk.Entry(root)
entry.pack()
entry.bind("<Key>", on_key_press)

# root.mainloop()  # Uncomment to run
```

## 4. Observable Variables (PCPP 3.3)

Link widgets to variables that automatically update.

```python
root = tk.Tk()

text_var = tk.StringVar(value="Hello")

entry = tk.Entry(root, textvariable=text_var, width=20)
entry.pack(pady=10)

label = tk.Label(root, textvariable=text_var)
label.pack()

# Variable automatically updates label when entry changes
# root.mainloop()  # Uncomment to run
```
