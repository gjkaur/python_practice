## 2. PEP 8: Style Guide

Code layout, naming, imports, and recommendations.

```python
# PEP 8: Naming conventions
CONSTANT_VALUE = 100  # UPPER_CASE for constants

def function_name():  # snake_case for functions
    variable_name = 42  # snake_case for variables
    return variable_name

class ClassName:  # PascalCase for classes
    def method_name(self):  # snake_case for methods
        pass
```

**Additional Example**: PEP 257 one-line docstring.


```python
def add(a, b):
    """Return sum of a and b."""
    return a + b

print(add.__doc__)
```

## 3. PEP 257: Docstrings

Documentation string conventions.

```python
def example_function(param1: int, param2: str) -> bool:
    """One-line docstring describing the function."""
    return True

def complex_function(param: int) -> dict:
    """Multi-line docstring with detailed description.
    
    This function demonstrates proper docstring format
    following PEP 257 guidelines.
    
    Args:
        param: Description of parameter
        
    Returns:
        Dictionary with results
        
    Raises:
        ValueError: If param is invalid
    """
    return {'result': param}
```

**Additional Example**: PEP 484 type hints.


```python
def discount(price: float, rate: float = 0.1) -> float:
    return price * (1 - rate)

print(discount(100.0))
```

## 4. PEP 484: Type Hints

Type annotations for function parameters and return values.

```python
from typing import List, Dict, Optional

def process_items(items: List[str]) -> Dict[str, int]:
    """Process list of items."""
    return {'count': len(items)}

def get_user(user_id: str) -> Optional[Dict[str, str]]:
    """Get user by ID, returns None if not found."""
    if user_id == '1':
        return {'id': '1', 'name': 'Alice'}
    return None
```
