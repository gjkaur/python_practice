## 1. Static Methods (PCPP 1.5)

Static methods are utility functions related to the class but don't need instance or class state.

```python
class MathUtils:
    @staticmethod
    def add(a: float, b: float) -> float:
        """Static method - no self or cls."""
        return a + b

print(f"MathUtils.add(5, 3): {MathUtils.add(5, 3)}")
# Can also call on instance
utils = MathUtils()
print(f"utils.add(10, 20): {utils.add(10, 20)}")
```

**Additional Example**: Static method as validator.


```python
class UserForm:
    @staticmethod
    def is_valid_email(email: str) -> bool:
        return '@' in email and '.' in email.split('@')[-1]

print(UserForm.is_valid_email('a@b.com'), UserForm.is_valid_email('bad'))
```

## 2. Class Methods (PCPP 1.5)

Class methods receive `cls` and can access/modify class state or create instances.

```python
class Date:
    def __init__(self, year: int, month: int, day: int):
        self.year = year
        self.month = month
        self.day = day
    
    @classmethod
    def from_string(cls, date_string: str):
        """Alternative constructor - factory method."""
        parts = date_string.split('-')
        return cls(int(parts[0]), int(parts[1]), int(parts[2]))
    
    def __str__(self):
        return f"{self.year}-{self.month:02d}-{self.day:02d}"

date = Date.from_string("2023-12-25")
print(f"Date: {date}")
```

**Additional Example**: Class method alternate constructor.


```python
class Temperature:
    def __init__(self, celsius: float):
        self.celsius = celsius
    @classmethod
    def from_fahrenheit(cls, f: float):
        return cls((f - 32) * 5 / 9)

print(Temperature.from_fahrenheit(32).celsius)
```

## 3. Abstract Base Classes (PCPP 1.6)

Abstract classes define interfaces that subclasses must implement.

```python
import abc

class Shape(abc.ABC):
    @abc.abstractmethod
    def area(self) -> float:
        """Abstract method - must be implemented."""
        pass
    
    @abc.abstractmethod
    def perimeter(self) -> float:
        pass

class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height
    
    def area(self) -> float:
        return self.width * self.height
    
    def perimeter(self) -> float:
        return 2 * (self.width + self.height)

rect = Rectangle(5, 3)
print(f"Rectangle area: {rect.area()}")

# Try to instantiate abstract class - raises TypeError
# shape = Shape()  # This would fail
```

**Additional Example**: Cannot instantiate ABC.


```python
import abc
class Port(abc.ABC):
    @abc.abstractmethod
    def send(self, data: bytes) -> None: ...

class MemoryPort(Port):
    def send(self, data: bytes) -> None:
        print('sent', len(data), 'bytes')

MemoryPort().send(b'hi')
```

## 4. Multiple Inheritance with ABCs (PCPP 1.6)

Classes can inherit from multiple abstract classes.

```python
class Drawable(abc.ABC):
    @abc.abstractmethod
    def draw(self) -> str:
        pass

class Movable(abc.ABC):
    @abc.abstractmethod
    def move(self, x: float, y: float):
        pass

class DrawableMovableShape(Shape, Drawable, Movable):
    def __init__(self, x: float, y: float):
        self.x = x
        self.y = y
    
    def area(self) -> float:
        return 0.0
    
    def perimeter(self) -> float:
        return 0.0
    
    def draw(self) -> str:
        return f"Drawing at ({self.x}, {self.y})"
    
    def move(self, x: float, y: float):
        self.x = x
        self.y = y

shape = DrawableMovableShape(10, 20)
print(shape.draw())
```

**Additional Example**: Multiple ABC mixins.


```python
import abc
class Serializable(abc.ABC):
    @abc.abstractmethod
    def to_dict(self) -> dict: ...

class User(Serializable):
    def __init__(self, name):
        self.name = name
    def to_dict(self):
        return {'name': self.name}

print(User('Ada').to_dict())
```
