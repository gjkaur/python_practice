## 1. Pickle Module (PCPP 1.11)

Serialize Python objects to byte streams.

```python
import pickle

class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age
    
    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"

person = Person("Alice", 30)

# Serialize to bytes
data = pickle.dumps(person)
print(f"Pickled: {data[:50]}...")

# Deserialize
restored = pickle.loads(data)
print(f"Restored: {restored}")
```

**Additional Example**: Pickle list of objects.


```python
import pickle
items = [{'id': 1}, {'id': 2}]
raw = pickle.dumps(items)
print('restored:', pickle.loads(raw))
```

## 2. Shelve Module (PCPP 1.11)

Dictionary-like interface to persistent storage.

```python
import shelve

# Write to shelve
with shelve.open("data.db", flag='c') as db:
    db['users'] = ["Alice", "Bob"]
    db['count'] = 2

# Read from shelve
with shelve.open("data.db", flag='r') as db:
    print(f"Users: {db['users']}")
    print(f"Count: {db['count']}")
```

**Additional Example**: Shelve update existing key.


```python
import shelve
from pathlib import Path
path = 'demo_shelf.db'
with shelve.open(path, flag='c') as db:
    db['counter'] = db.get('counter', 0) + 1
    print('counter:', db['counter'])
for ext in ('.db', '.bak', '.dat', '.dir'):
    p = Path(path + ext)
    if p.exists():
        p.unlink()
```

## 3. Metaclasses (PCPP 1.12)

Metaclasses are classes that create classes.

```python
class RegistryMeta(type):
    """Metaclass that registers classes."""
    registry = {}
    
    def __new__(mcs, name, bases, namespace):
        cls = super().__new__(mcs, name, bases, namespace)
        mcs.registry[name] = cls
        return cls

class MyClass(metaclass=RegistryMeta):
    pass

print(f"Registry: {RegistryMeta.registry}")
```

**Additional Example**: Metaclass auto __repr__.


```python
class AutoReprMeta(type):
    def __new__(mcs, name, bases, ns):
        cls = super().__new__(mcs, name, bases, ns)
        if name != 'Base':
            def __repr__(self):
                fields = ', '.join(f'{k}={v!r}' for k, v in self.__dict__.items())
                return f'{name}({fields})'
            cls.__repr__ = __repr__
        return cls

class Base(metaclass=AutoReprMeta):
    pass

class Point(Base):
    def __init__(self, x, y):
        self.x, self.y = x, y

print(Point(1, 2))
```

## 4. Dynamic Class Creation (PCPP 1.12)

Create classes at runtime using `type()`.

```python
# type(name, bases, dict)
DynamicClass = type('DynamicClass', (object,), {
    'x': 10,
    'get_x': lambda self: self.x
})

obj = DynamicClass()
print(f"Dynamic object: {obj.get_x()}")
```

**Additional Example**: type() with methods.


```python
def greet(self):
    return f'Hi, {self.name}'

Person = type('Person', (), {'name': 'Guest', 'greet': greet})
print(Person().greet())
```

## 5. Special Attributes (PCPP 1.12)

Access class metadata programmatically.

```python
class Example:
    pass

print(f"__name__: {Example.__name__}")
print(f"__bases__: {Example.__bases__}")
print(f"__dict__ keys: {list(Example.__dict__.keys())}")
print(f"__module__: {Example.__module__}")
```

**Additional Example**: __qualname__ for nested classes.


```python
class Outer:
    class Inner:
        pass

print(Outer.Inner.__qualname__)
```
