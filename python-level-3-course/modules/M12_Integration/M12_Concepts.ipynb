## Architecture Pattern: MVC

Model-View-Controller pattern for separation of concerns.

```python
# Model: Data layer (database)
class DatabaseModel:
    def get_data(self):
        return []

# Service: Business logic
class Service:
    def __init__(self, model):
        self.model = model
    
    def process_data(self):
        return self.model.get_data()

# View: GUI layer
class View:
    def __init__(self, service):
        self.service = service
    
    def display(self):
        data = self.service.process_data()
        print(data)
```

## Integration Checklist

When building integrated applications:

- [ ] Separate GUI from business logic
- [ ] Separate business logic from data access
- [ ] Use logging for important events
- [ ] Externalize configuration
- [ ] Handle errors gracefully
- [ ] Write tests for core logic
- [ ] Document architecture and usage
