## 1. Extended Function Arguments (PCPP 1.4)

`*args` collects positional arguments into a tuple, `**kwargs` collects keyword arguments into a dict.

```python
def sum_all(*args: float) -> float:
    """Accept variable number of positional arguments."""
    return sum(args)

print(f"sum_all(1, 2, 3): {sum_all(1, 2, 3)}")
print(f"sum_all(1, 2, 3, 4, 5): {sum_all(1, 2, 3, 4, 5)}")
```

```python
def create_user(name: str, email: str, **kwargs: dict) -> dict:
    """Accept variable keyword arguments."""
    user = {'name': name, 'email': email}
    user.update(kwargs)
    return user

user = create_user("Alice", "alice@example.com", age=30, city="NYC")
print(user)
```

**Additional Example**: Combining *args and **kwargs.


```python
def log_call(*args, **kwargs):
    print('args:', args)
    print('kwargs:', kwargs)

log_call(1, 2, mode='fast', debug=True)
```

## 2. Closures (PCPP 1.4)

Closures are inner functions that capture variables from their enclosing scope.

```python
def make_multiplier(n: int):
    """Function factory using closure."""
    def multiply(x: int) -> int:
        # Inner function has access to 'n' from outer scope
        return x * n
    return multiply

multiply_by_5 = make_multiplier(5)
print(f"multiply_by_5(10): {multiply_by_5(10)}")
```

**Additional Example**: Closure counter.


```python
def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc

c = make_counter()
print(c(), c(), c())
```

## 3. Function Decorators (PCPP 1.4)

Decorators modify or enhance functions. `@decorator` is syntactic sugar for `func = decorator(func)`.

```python
import functools
import time

def timer(func):
    """Decorator that measures execution time."""
    @functools.wraps(func)  # Preserve function metadata
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def slow_function():
    time.sleep(0.1)
    return "Done"

result = slow_function()
print(f"Result: {result}")
```

**Additional Example**: Simple @decorator without functools.


```python
def bold(fn):
    def wrapper(*a, **k):
        return f'**{fn(*a, **k)}**'
    return wrapper

@bold
def title():
    return 'Report'

print(title())
```

## 4. Decorators with Arguments (PCPP 1.4)

Three-level function pattern: outer takes args, middle takes func, inner is wrapper.

```python
def repeat(n: int):
    """Decorator factory - returns a decorator."""
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(n):
                result = func(*args, **kwargs)
            return result
        return wrapper
    return decorator

@repeat(3)
def greet(name: str):
    print(f"Hello, {name}!")
    return f"Greeted {name}"

result = greet("World")
```

**Additional Example**: Parameterized @repeat(n).


```python
import functools

def repeat(times):
    def deco(fn):
        @functools.wraps(fn)
        def wrapper(*a, **k):
            for _ in range(times):
                fn(*a, **k)
        return wrapper
    return deco

@repeat(2)
def ping():
    print('ping')

ping()
```

## 5. Class Decorators (PCPP 1.4)

Classes can act as decorators by implementing `__call__` method.

```python
class Logger:
    """Class decorator."""
    
    def __init__(self, module_name: str):
        self.module_name = module_name
    
    def __call__(self, func):
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            print(f"[{self.module_name}] Calling {func.__name__}")
            result = func(*args, **kwargs)
            print(f"[{self.module_name}] {func.__name__} completed")
            return result
        return wrapper

@Logger("math_utils")
def add(a: int, b: int) -> int:
    return a + b

result = add(5, 3)
```

**Additional Example**: Class decorator with state.


```python
class CountCalls:
    def __init__(self, fn):
        self.fn = fn
        self.calls = 0
    def __call__(self, *a, **k):
        self.calls += 1
        return self.fn(*a, **k)

@CountCalls
def work():
    return 'done'

work()
work()
print('calls:', work.calls)
```

## 6. Decorator Stacking

Decorators are applied bottom-to-top. Execution order: top-to-bottom.

```python
def decorator1(func):
    print("Decorator 1 applied")
    return func

def decorator2(func):
    print("Decorator 2 applied")
    return func

@decorator1
@decorator2
def my_function():
    pass

# Decorators applied: decorator2 first, then decorator1
```

**Additional Example**: Stacking order demo.


```python
def tag(name):
    def deco(fn):
        def wrapper(*a, **k):
            print(f'[{name}] enter')
            return fn(*a, **k)
        return wrapper
    return deco

@tag('outer')
@tag('inner')
def task():
    print('running')

task()
```
