## 1. JSON Serialization (PCPP 4.3)

Convert Python objects to/from JSON format.

```python
import json

# Python dict to JSON
data = {"name": "Alice", "age": 30}
json_string = json.dumps(data)
print(f"JSON: {json_string}")

# JSON to Python dict
parsed = json.loads(json_string)
print(f"Parsed: {parsed}")
```

**Additional Example**: JSON pretty print.


```python
import json
data = {'users': [{'name': 'Ada'}, {'name': 'Bob'}]}
print(json.dumps(data, indent=2))
```

## 2. XML Processing (PCPP 4.3)

Parse and create XML documents.

```python
import xml.etree.ElementTree as ET

# Parse XML
xml_string = """
<users>
    <user id="1">
        <name>Alice</name>
    </user>
</users>
"""

root = ET.fromstring(xml_string)
user = root.find('user')
name = user.find('name').text
print(f"User name: {name}")

# Create XML
new_root = ET.Element("products")
product = ET.SubElement(new_root, "product")
ET.SubElement(product, "name").text = "Widget"
ET.dump(new_root)
```

**Additional Example**: Build XML with attributes.


```python
import xml.etree.ElementTree as ET
root = ET.Element('catalog')
item = ET.SubElement(root, 'item', id='1')
ET.SubElement(item, 'name').text = 'Widget'
print(ET.tostring(root, encoding='unicode'))
```

## 3. REST Client with requests (PCPP 4.4)

Use requests library for HTTP communication.

```python
import requests

# GET request
response = requests.get("https://httpbin.org/get", params={"key": "value"})
print(f"Status: {response.status_code}")
print(f"Response: {response.json()}")

# POST request
post_data = {"name": "Alice", "age": 30}
response = requests.post("https://httpbin.org/post", json=post_data)
print(f"\nPOST Response: {response.json()}")
```

**Additional Example**: Handle HTTP errors.


```python
import requests
from requests.exceptions import HTTPError

try:
    r = requests.get('https://httpbin.org/status/404')
    r.raise_for_status()
except HTTPError as e:
    print('HTTP error:', e.response.status_code)
```
