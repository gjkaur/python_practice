## 1. OOP Terminology and Reflection (PCPP 1.1)

Essential terms: **class** (blueprint), **instance** (specific object), **object** (any instance), **attribute** (data), **method** (function on class/instance), **type** (class of an object).

```python
class Animal:
    """PCPP 1.1: Base class - the blueprint."""
    
    def __init__(self, name: str) -> None:
        self.name = name  # Instance variable
    
    def speak(self) -> str:
        """Method - behavior."""
        return "..."

# Create instances (objects)
a1 = Animal("Generic")
a2 = Animal("Other")

print(f"Type of a1: {type(a1)}")
print(f"a1 is instance of Animal: {isinstance(a1, Animal)}")
print(f"Animal is subclass of object: {issubclass(Animal, object)}")
print(f"a1.name (attribute): {a1.name}")
```

**Additional Example**: issubclass with built-in types.


```python
class Employee: pass
print(issubclass(Employee, object))
print(isinstance(Employee(), Employee))
```

## 2. Class Variables vs Instance Variables (PCPP 1.1)

**Instance variables**: unique to each instance. **Class variables**: shared by all instances (and the class).

```python
class Dog(Animal):
    """PCPP 1.1: Class variable vs instance variable."""
    
    species = "Canis familiaris"  # Class variable - shared
    
    def __init__(self, name: str, breed: str) -> None:
        super().__init__(name)
        self.breed = breed  # Instance variable - per object
    
    def speak(self) -> str:
        return "Woof!"

d1 = Dog("Max", "Labrador")
d2 = Dog("Bella", "Poodle")

print(f"d1.species: {d1.species}")  # Same for all dogs
print(f"d2.species: {d2.species}")
print(f"d1.breed: {d1.breed}")    # Different per instance
print(f"d2.breed: {d2.breed}")
print(f"Dog.species: {Dog.species}")  # Access via class
```

**Additional Example**: Mutating class variable affects all instances.


```python
class Tag:
    prefix = 'ID-'

a, b = Tag(), Tag()
Tag.prefix = 'REF-'
print(a.prefix, b.prefix)
```

## 3. Inheritance and MRO (PCPP 1.3)

**Single inheritance**: one parent. **Multiple inheritance**: more than one parent. **MRO** (Method Resolution Order) determines which method is called.

```python
class Cat(Animal):
    def speak(self) -> str:
        return "Meow!"

# Multiple inheritance example
class Flyable:
    def fly(self) -> str:
        return "Flying"

class Bird(Animal, Flyable):
    def speak(self) -> str:
        return "Tweet!"

# MRO: order Python looks for methods
print("Bird MRO:", [c.__name__ for c in Bird.__mro__])

b = Bird("Tweety")
print(b.speak())  # Bird.speak
print(b.fly())    # Flyable.fly
print(b.name)     # Animal.__init__
```

**Additional Example**: super() in multiple inheritance.


```python
class A:
    def __init__(self):
        print('A init')
class B(A):
    def __init__(self):
        super().__init__()
        print('B init')

B()
```

## 4. Polymorphism and Composition (PCPP 1.3)

**Polymorphism**: same interface (e.g. `speak()`), different implementations. **Composition**: "has a" relationship instead of "is a".

```python
# Polymorphism: same method name, different behavior
def make_it_speak(animal: Animal) -> None:
    print(animal.speak())  # Polymorphic call

make_it_speak(Dog("Max", "Lab"))
make_it_speak(Cat("Whiskers"))
make_it_speak(Bird("Tweety"))

# Composition: "has a" instead of "is a"
class Engine:
    def start(self) -> str:
        return "Engine started"

class Car:
    def __init__(self) -> None:
        self.engine = Engine()  # Car HAS-A Engine
    
    def start(self) -> str:
        return self.engine.start()

car = Car()
print(car.start())
```

**Additional Example**: Duck typing — no inheritance required.


```python
class Robot:
    def speak(self):
        return 'beep'

def announce(entity):
    print(entity.speak())

announce(Robot())
```
