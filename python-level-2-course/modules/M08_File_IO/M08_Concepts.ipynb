## 1. open() and Modes (PCAP 5.4, 5.5)

**open(path, mode="r", encoding="utf-8")** returns a **file object** (handle). You read from or write to it, then **close** it so the system releases the file. Prefer **with open(...) as f:** so the file is closed automatically when the block exits (even on exception).

**Modes**

- **"r"** — Read (default). File must exist. Text mode; use **encoding** for decoding bytes to str.
- **"w"** — Write. **Truncates** the file if it exists; creates it if not. Text mode.
- **"a"** — Append. Writes go to the end; file is created if missing. Text mode.
- **"rb"**, **"wb"** — Read/write in **binary** mode (bytes, no encoding). Use for images, etc.
- For **text** files, always pass **encoding="utf-8"** (or the correct encoding) so behavior is consistent across platforms. **Streams**: **sys.stdin**, **sys.stdout**, **sys.stderr** are pre-opened file-like objects.

```python
with open("tmp_demo.txt", "w", encoding="utf-8") as f:
    f.write("line1\nline2\n")
with open("tmp_demo.txt", encoding="utf-8") as f:
    print(f.read())
import os
if os.path.exists("tmp_demo.txt"):
    os.remove("tmp_demo.txt")
```

## 2. read, readline, readlines, write (PCAP 5.5)

**Reading**

- **f.read()** — Reads and returns the **entire** file as one string (text mode) or bytes (binary). For large files, this uses a lot of memory; consider **readline** or iterating over **f** line by line.
- **f.readline()** — Reads and returns **one line** (including the newline, if present). Returns **""** at end of file.
- **f.readlines()** — Reads all lines and returns a **list** of strings (each element is one line, with newline preserved). Alternatively, **list(f)** gives the same result.
- **for line in f:** — Iterates over the file **line by line** without loading the whole file into memory.

**Writing**

- **f.write(s)** — Writes the string **s** (text mode) or bytes (binary mode). Does **not** add a newline; add **"\\n"** yourself if needed. **f.writelines(sequence)** writes a sequence of strings (again, no automatic newlines).

**Errors**

- **OSError** (and its subclass **FileNotFoundError**) is raised when the file cannot be opened (e.g. missing file, permission denied). **e.errno** is the error code; **errno.ENOENT** means "no such file or directory." Use **try/except** and check **e.errno == errno.ENOENT** for missing files.

```python
import errno
try:
    open("nonexistent_xyz_123.txt")
except OSError as e:
    print("errno:", e.errno, "ENOENT:", errno.ENOENT, "same:", e.errno == errno.ENOENT)
```

```python
# read() vs readline() vs readlines(): create a small file and try each
with open("tmp_read_demo.txt", "w", encoding="utf-8") as f:
    f.write("first line\nsecond line\nthird line\n")

with open("tmp_read_demo.txt", encoding="utf-8") as f:
    whole = f.read()
    print("read():", repr(whole))

with open("tmp_read_demo.txt", encoding="utf-8") as f:
    lines = f.readlines()
    print("readlines():", lines)

with open("tmp_read_demo.txt", encoding="utf-8") as f:
    first = f.readline()
    second = f.readline()
    print("readline() x2:", repr(first), repr(second))
```

```python
# Append mode "a": add to the end without truncating
with open("tmp_append_demo.txt", "w", encoding="utf-8") as f:
    f.write("line1\n")
with open("tmp_append_demo.txt", "a", encoding="utf-8") as f:
    f.write("line2 (appended)\n")
with open("tmp_append_demo.txt", encoding="utf-8") as f:
    print("After append:", f.read())
```

## 4. Edge Cases and Best Practices

- **Always** use **with open(...) as f:** so the file is closed even if an exception occurs. Or call **f.close()** in a **finally** block.
- For **text** files, pass **encoding="utf-8"** (or the correct encoding) to avoid platform-dependent defaults.
- **OSError** (and **FileNotFoundError**, subclass) for missing file or permission; use **e.errno == errno.ENOENT** for "file not found".
- **bytearray** can be used as a mutable buffer for binary data; **bytes** is immutable.

## 5. More Examples

**Example: Read a file line by line with readline**

```python
with open("tmp.txt", "w", encoding="utf-8") as f:
    f.write("first\nsecond\nthird\n")
with open("tmp.txt", encoding="utf-8") as f:
    while True:
        line = f.readline()
        if not line:
            break
        print(line.rstrip())
```
