## 1. List Comprehensions (PCAP 5.1)

A **list comprehension** builds a new list in one expression by iterating over an iterable and optionally filtering.

- **[expr for x in iterable]** — Evaluates **expr** for each **x** in **iterable** and collects the results in a list. Equivalent to a **for** loop that appends to a list, but more concise.

- **[expr for x in iterable if condition]** — Only includes **expr** when **condition** is true. The **if** filters which elements are kept.

- **Nested**: **[expr for a in A for b in B]** — Like a nested **for** loop: for each **a** in **A**, iterate over **B**; **expr** can use both **a** and **b**. Order is left-to-right (first loop is the outer one). You can add **if** after the loops or after the inner loop.

- **Dict** and **set** comprehensions: **{k: v for ...}** and **{x for x in ...}** use the same idea for dictionaries and sets.

**Additional Example**: Filtered comprehension.


```python
words = ['apple', 'kiwi', 'banana', 'fig']
long_words = [w for w in words if len(w) >= 5]
print(long_words)
```

## 2. Lambdas and map/filter (PCAP 5.2)

- **lambda args: expr** is an anonymous function (single expression). Example: **lambda x: x + 1**.

- **map(f, iterable)** returns an iterator of **f(x)** for each **x**; **filter(pred, iterable)** keeps elements where **pred(x)** is true.

- **list(map(...))** or **list(filter(...))** to get a list.

```python
# List comprehension: squares of 0..4, then only even-indexed
squares = [x * x for x in range(5)]
print("Squares:", squares)
# With filter: only even numbers
evens_squared = [x * x for x in range(10) if x % 2 == 0]
print("Evens squared:", evens_squared)
# Nested: pairs (i, j) for i in 0,1 and j in 0,1
pairs = [(i, j) for i in range(2) for j in range(2)]
print("Pairs:", pairs)
```

```python
# Dict and set comprehensions
# Dict: {k: v for ...}
squares_dict = {x: x*x for x in range(5)}
print("Dict comprehension:", squares_dict)
# Set: {x for x in ...} — unique elements
chars = {c.lower() for c in "Hello"}
print("Set comprehension (unique chars):", chars)
```

**Additional Example**: sorted with key=lambda.


```python
people = [('Ada', 30), ('Bob', 25), ('Cy', 35)]
by_age = sorted(people, key=lambda p: p[1])
print(by_age)
```

## 3. Closures (PCAP 5.3)

- A **closure** is a function that "captures" variables from its enclosing scope. Example: **make_add(n)** returns **lambda x: x + n**; the returned function remembers **n**.

- The closed-over variable is read when the inner function **runs**, not when it is defined (important for loops: create closures in a function or use default arg to bind now).

**Additional Example**: Closure factory for validators.


```python
def min_length(n):
    def check(text):
        return len(text) >= n
    return check

validate = min_length(3)
print(validate('hi'), validate('hey'))
```

## 4. Edge Cases / 5. More Examples / 6. Practice

- **Dict/set comprehensions**: **{k: v for ...}**, **{x for x in ...}**.

- **Practice:** Build a list of squares of even numbers from 0 to 9 using a list comprehension. Use **filter(lambda x: x % 2 == 0, range(10))** and **map** to get the same result.

```python
# Lambda: anonymous function, single expression only
# lambda x: x + 1  — same as  def f(x): return x + 1
f = lambda x, y: x + y
print("lambda x, y: x + y  with (3, 4):", f(3, 4))
# Lambdas cannot have statements (no if/for/assign); only expressions
```

```python
# map and filter with lambda
data = [1, 2, 3, 4, 5]
incremented = list(map(lambda x: x + 1, data))
print("map(lambda x: x+1, data):", incremented)
positive = list(filter(lambda x: x > 2, data))
print("filter(lambda x: x>2, data):", positive)
# Same with list comprehensions:
print("Comprehension equivalent:", [x + 1 for x in data], [x for x in data if x > 2])
```

```python
evens_sq = [x*x for x in range(10) if x % 2 == 0]
print(evens_sq)
same = list(map(lambda x: x*x, filter(lambda x: x % 2 == 0, range(10))))
print(same == evens_sq)
```

```python
# Closure: the returned function "remembers" n
def make_add(n):
    return lambda x: x + n

add_2 = make_add(2)
add_10 = make_add(10)
print("add_2(3):", add_2(3))
print("add_10(3):", add_10(3))
# Loop pitfall: without default arg, all refer to same i
funcs_bad = [lambda: i for i in range(3)]
funcs_good = [lambda i=i: i for i in range(3)]
print("Without default (all see final i):", [f() for f in funcs_bad])
print("With default (each captures its i):", [f() for f in funcs_good])
```
