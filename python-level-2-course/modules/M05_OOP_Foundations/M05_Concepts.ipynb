## 1. Class and Object (PCAP 4.1)

**Classes** and **objects** are the basis of object-oriented programming in Python.

- **Class** — A **blueprint** or template. You define it with **class Name:** and indented attributes and methods. It describes what data (attributes) and behavior (methods) each **object** of that type will have.
- **Object** — An **instance** of a class. You create one by **calling** the class like a function: **obj = MyClass()**. Each object has its own copy of instance attributes (e.g. **obj.name**) and shares the class's methods.
- **Properties / attributes** — Data stored on an object (e.g. **self.name**, **self.price**). Accessed as **obj.name**.
- **Methods** — Functions defined inside the class that operate on the object. The first parameter is **self**, which refers to the instance. Called as **obj.method(args)**.
- **Encapsulation** — Bundling data and the operations on that data into one unit (the class). Code that uses the object only needs to call methods instead of manipulating internal state directly.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def format_line(self):
        return f"{self.name}: {self.price:.2f}"
p = Product("Widget", 9.99)
print(p.format_line())
```

## 2. Instance vs Class Variables (PCAP 4.2)

**Instance variables** — Defined and assigned on **self** (e.g. in **__init__**): **self.name = name**. Each **object** has its own copy. Used for data that differs per instance (e.g. each Product has its own name and price).

**Class variables** — Defined directly under the **class** (e.g. **count = 0**). Shared by **all** instances of the class. Used for constants, default values, or shared state (e.g. a counter of how many instances were created). Access as **Class.name** or **obj.name** (lookup falls back to the class if not on the instance).

**__dict__** — **obj.__dict__** is a dictionary of the object's **instance** attributes (names and values). **Class.__dict__** (or **vars(Class)**) contains the class's attributes (methods, class variables, etc.). Useful for debugging and introspection.

**Private** — A single leading underscore (e.g. **_helper**) is a **convention** meaning "internal use." Double leading underscore (e.g. **__name**) triggers **name mangling** in classes (the name is rewritten to include the class name), used to avoid accidental overrides in subclasses.

```python
class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
        self.id = Counter.count
c1, c2 = Counter(), Counter()
print(c1.id, c2.id, Counter.count)
print("c1.__dict__:", c1.__dict__, "Counter.__dict__ keys:", list(Counter.__dict__.keys())[:5])
```

```python
# Instance __dict__ vs class __dict__: instance has only its own attributes
print("c1.__dict__ (instance):", c1.__dict__)
print("Counter.__dict__ (class) - keys:", list(Counter.__dict__.keys()))
# Class variable Counter.count is shared; c1.id and c2.id are per-instance
print("Counter.count (class var):", Counter.count)
```

## 3. Methods and self (PCAP 4.3)

**Instance methods** — The **first parameter** must be **self**; when you call **obj.method(args)**, Python passes **obj** as **self**. Inside the method, **self** refers to that instance. **__init__(self, ...)** is the **constructor**; it is called automatically when you create an object and must not return a value (other than None).

```python
class Account:
    def __init__(self, name, balance=0):
        self.name = name
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        return self.balance
a = Account("Alice", 100)
print(a.deposit(50), a.balance)
```

## 4. Introspection (PCAP 4.4)

**hasattr(obj, name)**: True if object has that attribute. **Class attributes**: __name__, __module__, __bases__ (base classes). Use for debugging and generic code.

```python
print(hasattr(a, "balance"), hasattr(Account, "deposit"))
print("Account.__name__:", Account.__name__, "Account.__module__:", Account.__module__, "Account.__bases__:", Account.__bases__)
```

---
## 5. Constructors (PCAP 4.6)

- **__init__(self, ...)** is the usual constructor; called automatically when the object is created.
- Must not return anything other than **None** (Python ignores any return value).
- Use it to assign instance attributes and light validation; avoid heavy I/O or complex logic.

---
## 6. Built-in and Related / Edge Cases

- **hasattr(obj, name)**, **getattr(obj, name [, default])**, **setattr(obj, name, value)** – attribute access by string.
- **vars(obj)** is like **obj.__dict__** for objects with a __dict__.
- **type(obj)** gives the class; **isinstance(obj, Class)** for type checks (see M06).
- **Edge cases**: Mutating a class variable (e.g. list.append) affects all instances; rebinding (Class.x = ...) affects the class. Forgetting **self** as first parameter in instance methods causes errors. **__init__** must not return a value.

## 7. More Examples

**Example: Class variable as shared counter**

```python
class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
        self.id = Counter.count
c1, c2, c3 = Counter(), Counter(), Counter()
print(c1.id, c2.id, c3.id, Counter.count)
```
