## 1. Encoding and ord/chr (PCAP 3.1, 3.2)

### Concepts

Text in Python 3 is represented as **Unicode** strings. Each character has a numeric **code point** (e.g. 65 for 'A', 233 for 'é').

- **ASCII** — A 7-bit encoding; code points 0–127. Covers basic Latin letters, digits, and control characters. Limited to English and a few symbols.

- **Unicode** — A standard that assigns a **code point** (integer) to almost every character in every script (Latin, Greek, Cyrillic, CJK, emoji, etc.). Written as U+0041 for 'A'.

- **UTF-8** — A **variable-length** encoding of Unicode: each code point is stored as 1–4 bytes. Code points 0–127 are stored as one byte, same as ASCII, so ASCII text is valid UTF-8. UTF-8 is the default in most modern systems and in Python when you open files with **encoding="utf-8"**.

- **ord(c)** — Takes a **single-character** string and returns its Unicode code point (int). **chr(i)** does the reverse: returns the character for code point **i** (0 ≤ i ≤ 0x10FFFF).

- **Escape sequences** — Inside string literals, **\n** is newline, **\t** is tab, **\\** is backslash, **\uXXXX** is a 16-bit hex code point, **\UXXXXXXXX** is 32-bit, **\N{name}** is by Unicode name.

```python
print(ord("A"), chr(65))
print(ord("e"), chr(233))
print("Hello\\nWorld")
```

```python
# Escape sequences and Unicode by name
print("Tab\tbetween\twords")
print("Backslash: \\")
print("Unicode by name: \N{EURO SIGN}", "\N{GREEK CAPITAL LETTER ALPHA}")
# Code point for Euro and Alpha
print("ord:", ord("\N{EURO SIGN}"), ord("\N{GREEK CAPITAL LETTER ALPHA}"))
```

```python
# ord("") raises TypeError — empty string has no code point
try:
    ord("")
except TypeError as e:
    print("ord('') raises TypeError:", e)
# chr() works for valid code points
print("chr(128512):", chr(128512))  # emoji
```

**Additional Example**: ord and chr.


```python
print('ord A:', ord('A'), 'chr 65:', chr(65))
```

## 2. Indexing, Slicing, Immutability (PCAP 3.2)

**Indexing**

- **s[i]** — Returns the character at index **i** (0-based). **s[0]** is the first character. **Negative** indices count from the end: **s[-1]** is the last character, **s[-2]** the second-to-last. Raises **IndexError** if **i** is out of range.

**Slicing**

- **s[start:stop]** — Returns the substring from index **start** up to (but not including) **stop**. So **s[1:4]** is characters at indices 1, 2, 3. If you omit **start**, it defaults to 0; if you omit **stop**, it defaults to **len(s)**. Unlike indexing, slicing **never** raises **IndexError**: out-of-range or "backward" ranges yield an empty string or a shorter substring.

- **s[start:stop:step]** — Same as above but takes every **step**-th character. **step** can be negative: **s[::-1]** means "all characters in reverse" (step -1, default start/stop reversed), so it **reverses** the string.

**Immutability**

- Strings are **immutable**: you cannot change a character in place (e.g. **s[0] = "x"** is invalid). Methods like **.upper()** or **.replace()** return a **new** string; they do not modify the original.

**Membership and iteration**

- **sub in s** / **sub not in s** — True if **sub** appears as a substring. **for c in s:** iterates over each **character** in **s**.

```python
s = "Hello, World"
print(s[0], s[-1], s[1:5], s[::-1][:5])
print("ell" in s, "xyz" not in s)
print([c for c in s[:3]])
```

```python
# Step in slicing: s[::2] = every 2nd char; s[::-1] = reverse
t = "Python"
print("Every 2nd character:", t[::2])
print("Reverse:", t[::-1])
# Immutability: .upper() returns NEW string; original unchanged
u = "hello"
u2 = u.upper()
print("Original:", u, "After .upper():", u2)
# u[0] = "H"  # would raise TypeError: 'str' object does not support item assignment
```

**Additional Example**: String immutability.


```python
s = 'hello'
# s[0] = 'H'  # TypeError
upper = s.upper()
print('original:', s, '| upper copy:', upper)
```

## 3. String Methods (PCAP 3.3)

### Predicate methods (.isxxx())

| Method | True when |
|--------|-----------|
| **.isalpha()** | All characters are letters |
| **.isdigit()** | All characters are decimal digits |
| **.isalnum()** | All characters are alphanumeric |
| **.isspace()** | All characters are whitespace |

### Search and split/join

| Method | Description | Returns / Notes |
|--------|-------------|-----------------|
| **.find(sub)** | Index of first occurrence | -1 if not found |
| **.rfind(sub)** | Index of last occurrence | -1 if not found |
| **.index(sub)** | Like find | **ValueError** if not found |
| **.split(sep=None)** | Split by sep (default: whitespace) | list of strings |
| **sep.join(iterable)** | Join strings with sep (call on separator) | str |

**sorted(s)** returns a sorted list of characters (one per element).

```python
# sorted(s): returns list of characters in sorted order
s = "hello"
print("sorted('hello'):", sorted(s))
print("''.join(sorted(s)):", "".join(sorted(s)))
# str vs repr: repr adds quotes and escapes
x = "a\nb"
print("str:", str(x), "| repr:", repr(x))
```

```python
print("  ".join(["a","b","c"]))
print("a b c".split())
print("hello".find("l"), "hello".rfind("l"))
print("ab12".isalnum(), "12".isdigit())
try:
    "hello".index("x")
except ValueError as e:
    print("index('x') raises ValueError:", e)
```

```python
# .isxxx() methods: check character categories
for s in ["abc", "123", "abc123", "   ", "a1!", ""]:
    print(repr(s), "isalpha:", s.isalpha(), "isdigit:", s.isdigit(), "isalnum:", s.isalnum(), "isspace:", s.isspace())
```

```python
# .find() vs .index(): find returns -1 if not found; index raises ValueError
s = "hello world"
print("find('o'):", s.find("o"), "rfind('o'):", s.rfind("o"))
print("find('xyz'):", s.find("xyz"))
try:
    s.index("xyz")
except ValueError as e:
    print("index('xyz') raises ValueError:", e)
```

**Additional Example**: encode/decode UTF-8.


```python
text = 'café'
b = text.encode('utf-8')
print('bytes:', b)
print('decoded:', b.decode('utf-8'))
```

## 5. Common Errors and Edge Cases

- **ord("")** raises **TypeError** (empty string has no code point).

- **.index(sub)** raises **ValueError** when sub not found; use **.find()** if you want -1.

- When reading/writing files, pass **encoding="utf-8"** (or the correct encoding) explicitly.

- Building long strings in a loop with **+=** is inefficient; prefer a list of parts then **"".join(parts)**.

```python
# Efficient string building: use list + join instead of += in a loop
# BAD: result = ""; for x in items: result += str(x)  # O(n^2)
# GOOD:
parts = [str(i) for i in range(5)]
result = "".join(parts)
print("join(parts):", result)
```

**Additional Example**: isalpha / isdigit checks.


```python
tokens = ['abc', '123', 'a1']
for t in tokens:
    print(t, 'alpha:', t.isalpha(), 'digit:', t.isdigit())
```

## 6. More Examples

**Example: Reverse word order with split and join**

```python
phrase = "one two three"
words = phrase.split()
reversed_phrase = " ".join(reversed(words))
print(reversed_phrase)
```

**Example: find vs index – safe lookup**

```python
s = "hello world"
idx = s.find("xyz")  # -1 if not found
if idx >= 0:
    print("Found at", idx)
else:
    print("Not found")
# s.index("xyz") would raise ValueError
```
