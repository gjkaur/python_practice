## 1. Single Inheritance and Overriding (PCAP 4.5)

**Inheritance** lets you define a new class based on an existing one. The new class (**subclass**) gets all attributes and methods of the existing class (**superclass**) and can add or change behavior.

- **Syntax**: **class Dog(Animal):** — **Dog** is a **subclass** of **Animal**; it **inherits** all of **Animal**'s methods and class attributes. You can add new methods or attributes and **override** existing methods.

- **Overriding** — If the subclass defines a method with the **same name** as the superclass, the subclass version is used when you call it on a subclass instance. So **Dog.speak()** returns **"Woof"** even though **Animal** defined **speak()** to return **"?"**. The subclass **replaces** (or extends) the behavior.

- **Polymorphism** — Code that expects an **Animal** can accept a **Dog** (or any subclass) because a **Dog** **is an** **Animal**. You can write functions that take **Animal** and call **animal.speak()**; they work for **Dog**, **Cat**, etc., and each type's overridden method runs. That's **polymorphism**: one interface, different behavior per type.

## 2. isinstance and Polymorphism

- **isinstance(obj, Class)**: True if **obj** is an instance of **Class** or a subclass. Use for type checks without breaking polymorphism.

- **type(obj)** returns the exact class; **isinstance** allows subclasses. Prefer **isinstance** when you want "any Animal".

```python
# Polymorphism: a function that accepts any Animal works with Dog (or any subclass)
def greet(animal):
    return animal.speak()

print(greet(Dog()))
print(greet(Animal()))
# isinstance: check if an object is an instance of a class (or subclass)
d = Dog()
print("d is Dog:", isinstance(d, Dog))
print("d is Animal:", isinstance(d, Animal))
print("type(d):", type(d))
```

```python
# type() vs isinstance(): type gives exact class; isinstance accepts subclasses
d = Dog()
print("type(d):", type(d))           # <class '__main__.Dog'>
print("type(d) == Dog:", type(d) == Dog)
print("type(d) == Animal:", type(d) == Animal)  # False — Dog is not exactly Animal
print("isinstance(d, Dog):", isinstance(d, Dog))
print("isinstance(d, Animal):", isinstance(d, Animal))  # True — Dog subclasses Animal
```

## 3. super() and __str__

- **super()** calls the superclass implementation (e.g. **super().__init__(...)**, **super().method()**). Use to extend rather than replace.

- **__str__(self)** should return a readable string; used by **str(obj)** and **print(obj)**. Override for user-friendly output.

```python
# super() in __init__: call the superclass constructor, then add subclass state
class Person:
    def __init__(self, name):
        self.name = name
    def __str__(self):
        return f"Person({self.name})"

class Student(Person):
    def __init__(self, name, sid):
        super().__init__(name)  # must call super first to set self.name
        self.sid = sid
    def __str__(self):
        return f"Student({self.name}, sid={self.sid})"

s = Student("Alice", 101)
print(s)  # uses Student.__str__
print(str(s))
```

```python
# super() to extend a method (not just __init__): call superclass, then add more
class Greeter:
    def greet(self):
        return "Hello"
class GreeterEx(Greeter):
    def greet(self):
        return super().greet() + " — extended!"
g = GreeterEx()
print(g.greet())  # "Hello — extended!"
```

```python
class Person:
    def __init__(self, name):
        self.name = name
    def __str__(self):
        return f"Person({self.name})"
class Student(Person):
    def __init__(self, name, sid):
        super().__init__(name)
        self.sid = sid
    def __str__(self):
        return f"Student({self.name}, {self.sid})"
s = Student("Alice", 101)
print(s)
```

```python
# Multiple inheritance and MRO: class C(A, B) — method resolution order
class A:
    def method(self):
        return "A"
class B:
    def method(self):
        return "B"
class C(A, B):
    pass
c = C()
print("c.method():", c.method())  # Uses A's method (A comes first in C(A, B))
print("C.__mro__:", [x.__name__ for x in C.__mro__])
```

## 4. Multiple Inheritance and MRO

- **class C(A, B):** inherits from both; **method resolution order (MRO)** determines which method is used (e.g. **C.__mro__**).

- **Diamond** problem: A, B both inherit from Base, C inherits from A and B. Python uses **C3 linearization**; check **ClassName.__mro__**.

```python
# is vs ==: is checks identity (same object); == checks value
a = [1, 2, 3]
b = [1, 2, 3]
print("a == b:", a == b)   # True — same values
print("a is b:", a is b)   # False — different objects
c = a
print("a is c:", a is c)   # True — same object
# Prefer "if x is None" instead of "if x == None"
x = None
print("x is None:", x is None)
```
